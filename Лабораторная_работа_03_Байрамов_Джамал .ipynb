## Задание

Напишите функцию, которая вычисляет и возвращает факториал натурального числа, переданного функции в качестве параметра.

## Решение

### Объявление функции

```python
def factorial(n):
    """
    Возвращает факториал натурального числа n.
    n! = 1 * 2 * ... * n
    """
    if not isinstance(n, int):
        raise TypeError("n должно быть целым числом")
    if n < 0:
        raise ValueError("n должно быть неотрицательным числом")

    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
```

### Демонстрация использования функции

```python
# Демонстрация работы функции на нескольких значениях
numbers = [0, 1, 2, 5, 7, 10]

for n in numbers:
    print(f"{n}! = {factorial(n)}")
```

# Контрольные вопросы
**1. Чем отличается функция с возвратом значения от функций без возврата значения?**
Функция с возвратом значения после выполнения возвращает результат через `return`. Этот результат можно сохранить в переменную или использовать в выражении. Функция без возврата значения ничего не возвращает (фактически возвращает `None`) — она просто выполняет какие-то действия, например, печатает текст на экране.

**2. Что такое библиотечная функция?**
Это функция, которая уже написана разработчиками Python или сторонних библиотек. Чтобы её использовать, нужно подключить нужный модуль (например, `import random`). Самому писать такую функцию не нужно — она уже готова.

**3. Почему библиотечные функции похожи на «черные ящики»?**
Потому что мы знаем, *что* функция делает и какие аргументы ей нужны, но не видим, *как* именно она устроена внутри. Мы просто вызываем её и получаем результат, не вникая в реализацию.

**4. Что делает инструкция `x = random.randint(1, 100)`?**
Она вызывает функцию `randint` из модуля `random`, которая возвращает случайное целое число от 1 до 100 включительно. Это число записывается в переменную `x`.

**5. Что делает инструкция `print(random.randint(1, 20))`?**
Она генерирует случайное целое число от 1 до 20 и сразу выводит его на экран.

**6. Для чего используется вызов `random.seed(20)`?**
Он задаёт начальное значение (зерно) для генератора случайных чисел. Это нужно, чтобы последовательность случайных чисел была одинаковой при каждом запуске программы. Если seed один и тот же, то и случайные числа будут те же.

**7. Каким образом функции помогают повторно использовать код?**
Мы один раз пишем функцию, а потом можем вызывать её много раз в разных местах программы с разными аргументами. Не нужно копировать один и тот же код заново.

**8. Назовите и опишите две части определения функции.**

**Заголовок** — строка `def имя_функции(параметры):`. Здесь указывается имя функции и её параметры.

**Тело** — блок инструкций с отступом. Это то, что функция выполняет. Также может быть строка документации (описание функции).

**9. Как у функции задать параметры по умолчанию?**
В заголовке функции после имени параметра пишется знак `=` и значение по умолчанию. Например: `def func(a, b=10):`. Если при вызове второй аргумент не передан, то `b` автоматически станет равным 10.

**10. Что такое рекурсия?**
Это приём, когда функция вызывает саму себя. Обычно используется для задач, которые можно разбить на более простые подзадачи того же типа (например, вычисление факториала или обход дерева). Важно, чтобы было условие выхода из рекурсии, иначе она будет бесконечной.
