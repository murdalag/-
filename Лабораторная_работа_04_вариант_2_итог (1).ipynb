## Задание

В файле data_v2.txt посчитайте количество символов, расположенных между чётными цифрами.

## Решение

Скопируем файл data_v2.txt в локальное хранилище Google Colab (файл можно также загрузить в Colab вручную через панель «Файлы»):

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/data_v2.txt
````

Считываем весь файл в строку `s` методами `open()` и `read()`. Метод `strip()` убирает лишние пробелы и переводы строк по краям. Чтение обёрнуто в `try/except` на случай, если файл не найден.

````python
try:
    with open('data_v2.txt') as f:
        s = f.read().strip()
except FileNotFoundError:
    print('Файл data_v2.txt не найден')
    s = ''

print(len(s))
````

Обходим строку по индексу, пропуская первый и последний символы (у них нет двух соседей). Символ `s[i]` считается расположенным между чётными цифрами, если и слева `s[i-1]`, и справа `s[i+1]` от него стоят чётные цифры (0, 2, 4, 6, 8). Считается любой символ посередине — буква или цифра.

````python
even = '02468'   # чётные цифры
count = 0

for i in range(1, len(s) - 1):
    if s[i - 1] in even and s[i + 1] in even:
        count += 1

print('Количество символов между чётными цифрами =', count)
````

## Ответы на контрольные вопросы

**1.** Функции читают данные из текстового файла, открытого для чтения:
- `read()` считывает всё содержимое файла (или указанное число символов) в одну строку;
- `readline()` считывает одну строку файла (вместе с символом конца строки `\n`) и возвращает её как строку;
- `readlines()` считывает все строки файла и возвращает их списком строк.

Сходство: все три читают текст из файла и двигают позицию считывания вперёд. Различие — в объёме и типе результата: `read()` даёт одну строку со всем текстом, `readline()` — одну строку файла, `readlines()` — список всех строк.

**2.** Можно обойти файл циклом `for`: файловый объект является итерируемым, и на каждой итерации цикл получает очередную строку файла:
```python
with open('file.txt') as f:
    for line in f:
        print(line)
```

**3.** Закрытие файла завершает работу с ним: сбрасывает данные из буфера на диск (при записи), освобождает ресурсы операционной системы (файловый дескриптор) и разрешает другим программам работать с файлом. Если файл не закрыть, данные могут не записаться или файл останется занятым. Конструкция `with open(...)` закрывает файл автоматически.

**4.** Позиция считывания — это место в файле, начиная с которого будет прочитан следующий символ (или строка). При открытии файла для чтения позиция считывания находится в начале файла. По мере чтения она сдвигается вперёд.

**5.** Для дозаписи файл открывается в режиме `'a'` (append). Новые данные записываются в конец файла, существующее содержимое остаётся нетронутым.

**6.** Исключение — это ошибка, возникающая во время выполнения программы, которая прерывает обычный ход выполнения (например, деление на ноль, открытие несуществующего файла).

**7.** Если исключение вызвано и не обработано инструкцией `try/except`, программа аварийно завершается, а интерпретатор выводит сообщение об ошибке (трассировку стека).

**8.** При попытке открыть несуществующий файл (в режиме чтения) вызывается исключение `FileNotFoundError` (подвид `OSError`).

**9.** При попытке преобразовать нечисловую строку функцией `float()` вызывается исключение `ValueError`.

**10.** Если файл не существует и открывается в режиме дозаписи (`'a'`), ошибки не возникает: файл будет создан, а данные записаны в него.
