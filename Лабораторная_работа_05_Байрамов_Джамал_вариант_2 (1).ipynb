## Задание

В файле pr5.txt содержится последовательность целых чисел. Элементы последовательности могут принимать целые значения от –10 000 до 10 000 включительно. Определите количество пар, в которых хотя бы один из двух элементов больше, чем среднее арифметическое всех чисел в файле, и хотя бы один из двух элементов делится на 17. В ответе запишите два числа: сначала количество найденных пар, а затем – максимальную сумму элементов таких пар. В данной задаче под парой подразумевается два идущих подряд элемента последовательности. При решении задачи необходимо использовать списочный тип данных.

## Решение

Скопируем файл pr5.txt в локальное хранилище Google Colab:

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5.txt
````

Считываем числа из файла в список `data`, вычисляем среднее арифметическое и перебираем все пары соседних элементов `data[i]` и `data[i+1]`. Условие «хотя бы один» записано через `or`, оба условия объединены через `and`. Делимость на 17 проверяется как `% 17 == 0`.

````python
# Создаем пустой список и заполняем его числами из файла
data = []
with open('pr5.txt') as F:
    for line in F:
        data.append(int(line))

print(len(data))
print(data[:10])

# Среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)
print(sr)

counter = 0       # количество подходящих пар
max_sum = None    # максимальная сумма элементов таких пар

for i in range(len(data) - 1):
    # пара - два идущих подряд элемента
    x, y = data[i], data[i + 1]
    # хотя бы один больше среднего И хотя бы один делится на 17
    if (x > sr or y > sr) and (x % 17 == 0 or y % 17 == 0):
        counter += 1
        s = x + y
        if max_sum is None or s > max_sum:
            max_sum = s

print(f'Количество подходящих пар = {counter}')
print(f'Максимальная сумма = {max_sum}')
print(f'Ответ: {counter} {max_sum}')
````

**Ответ:** 881 18941

Вариант а) `names(0) = 'Вэнди'` — синтаксически неверен: круглые скобки означают вызов функции, а не обращение к элементу (для индексации используются квадратные скобки). Но даже запись `names[0] = 'Вэнди'` не подошла бы: список `names` пуст, и присваивание по несуществующему индексу вызовет `IndexError`. Метод `append()` добавляет элемент в конец списка, и в пустом списке он как раз окажется в позиции 0.

**5.** Списковые методы:

 а) `index(x)` — возвращает индекс первого элемента списка, равного `x`; если элемента нет, возникает `ValueError`;

 б) `insert(i, x)` — вставляет элемент `x` в позицию `i`, сдвигая остальные элементы вправо;
 в) `sort()` — сортирует список на месте (по возрастанию; для обратного порядка параметр `reverse=True`);

 г) `reverse()` — переворачивает список на месте, меняя порядок элементов на обратный.
